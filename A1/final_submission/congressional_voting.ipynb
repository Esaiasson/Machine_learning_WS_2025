Unique values in predictors are: \
'y', 'n', 'unknown'

For decision tree and random forest classifiers, the dataset can work as it is but since for knn we need to encode 'unknown' as 0.5, so that 'unknown' distance from  'y' and 'n' is equal and the information the the person abstaining to answer is preserved, preprocessing is applied for all.

## Feature plots

```python
# Compute normalized value counts for the 'class' column
target_counts = (
    congress_df_train['class']
    .value_counts(normalize=True)
    .reindex(['democrat', 'republican'])  # ensure consistent order
)

# Define colors
color_map = {'democrat': 'red', 'republican': 'blue'}

# Create horizontal stacked bar
fig = go.Figure()

start = 0
for party, proportion in target_counts.items():
    fig.add_trace(go.Bar(
        y=['Class Distribution'],       # single bar
        x=[proportion],
        name=party.capitalize(),
        orientation='h',
        marker=dict(color=color_map[party]),
        hovertemplate=f"{party}: " + "%{x:.1%}<extra></extra>"
    ))

fig.update_layout(
    barmode='stack',
    xaxis_title='Proportion',
    xaxis_tickformat='.0%',
    yaxis_title='',
    title='Class Distribution (Democrat vs Republican)',
    showlegend=True,
    template='plotly_white',
    width=700,
    height=250,
    margin=dict(l=80, r=40, t=60, b=40)
)

fig.show()


cols = congress_df_train.columns[1:]

# Define colors for feature categories
feature_color_map = {'y': 'green', 'unknown': 'darkgrey', 'n': 'crimson'}

# Compute normalized counts for all features
props = congress_df_train[cols].apply(lambda x: x.value_counts(normalize=True)).T

# Create horizontal stacked bar

fig = go.Figure()

for category in props.columns:
    fig.add_trace(go.Bar(
        y=props.index,                # feature names
        x=props[category],            # normalized values
        name=str(category),
        orientation='h',
        marker=dict(color=feature_color_map.get(category)),
        hovertemplate='%{y}<br>%{x:.2%}<extra>' + str(category) + '</extra>'
    ))

fig.update_layout(
    barmode='stack',
    xaxis_title='Proportion',
    yaxis_title='Feature',
    xaxis_tickformat='.0%',
    title='Feature Distributions',
    legend_title_text='Category',
    height=600,
    width=900,
    template='plotly_white',
    margin=dict(l=120, r=50, t=60, b=40)
)

fig.show()
```

## Preprocessing

### Custom Label Encoding

```python

def label_encoding(mapping, df):
    df_encoded = df.copy()
    for col in df_encoded.columns:
        unique_vals = set(df_encoded[col].unique())
        # Check if all unique values are in mapping keys
        if unique_vals.issubset(mapping.keys()):
            df_encoded[col] = df_encoded[col].map(mapping)
        else:
            print(f"Skipping column '{col}' — unexpected values found: {unique_vals - set(mapping.keys())}")
    return df_encoded
```

# Train Prep

```python
# define mappings
mapping_X = {
    'y': 1, 
    'unknown': 0.5,
    'n': 0
    }

# do the encoding
congress_preprocessed_train = label_encoding(mapping_X, congress_df_train)

# export to csv
congress_preprocessed_train.to_csv("congressional_df_train_preprocessed.csv", index=False)

congress_preprocessed_train.head()
```
