```python
import pandas as pd
```

```python
mcintire_news = pd.read_csv("./data/fake_and_real_news_dataset.csv")
```

```python
mcintire_news.head()
```

```python
def train_test_split(df):
    
    train_df = df.sample(frac=2/3, random_state=42)
    test_df = df.drop(train_df.index)
    
    return train_df, test_df
```

```python
mcintire_news_train, mcintire_news_test = train_test_split(mcintire_news)
```
